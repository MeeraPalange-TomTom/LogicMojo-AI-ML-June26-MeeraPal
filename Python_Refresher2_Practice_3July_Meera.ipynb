# Program to check Palindrome

```python
s = input("Enter a string or number: ")
cleaned = s.lower().replace(" ", "")
print("Palindrome" if cleaned == cleaned[::-1] else "Not Palindrome")
```

```python
s = input("Enter a string: ")
print("Palindrome" if s == s[::-1] else "Not Palindrome")
```

# Program to count number of vowels and print all the vowels

```python
s = input("Enter a string: ")
vowels = [c for c in s if c.lower() in "aeiou"]
print("Count:", len(vowels))
print("Vowels:", vowels)
```
